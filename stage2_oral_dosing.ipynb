# Stage 2: Oral dosing with first order absorption

Now the drug is swallowed, not injected. It is absorbed into the blood at a first order rate, ka, and it is eliminated at rate k at the same time. The curve rises first, then falls.

## Teaching example parameters, not a real drug

Dose = 100 mg, V = 50 L, CL = 5 L per hour, ka = 1.0 per hour, bioavailability F = 1.0. Teaching numbers only. They are not for any real drug and not for any real patient.

```python
import numpy as np
import matplotlib.pyplot as plt

# Teaching example parameters only
dose_mg = 100.0
V_L = 50.0
CL_L_per_h = 5.0
ka_per_h = 1.0
F = 1.0
k_per_h = CL_L_per_h / V_L

def oral_concentration(t):
    return (F * dose_mg * ka_per_h / (V_L * (ka_per_h - k_per_h))) * (np.exp(-k_per_h * t) - np.exp(-ka_per_h * t))

t = np.linspace(0, 48, 400)
C = oral_concentration(t)
Cmax = float(np.max(C))
Tmax = float(t[np.argmax(C)])
Tmax_formula = np.log(ka_per_h / k_per_h) / (ka_per_h - k_per_h)
print('Cmax from the curve:', round(Cmax, 2), 'mg per L')
print('Tmax from the curve:', round(Tmax, 2), 'hours')
print('Tmax from the formula:', round(Tmax_formula, 2), 'hours')
```

```python
plt.figure()
plt.plot(t, C)
plt.plot([Tmax], [Cmax], 'o')
plt.xlabel('Time (hours)')
plt.ylabel('Concentration (mg per L)')
plt.title('Oral one compartment model, teaching example')
plt.grid(True)
plt.show()
```
