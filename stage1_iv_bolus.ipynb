# Stage 1: One compartment IV bolus model

We start with the simplest PK model. One dose is given straight into the blood, and the drug leaves the body at a rate that depends on how much is there.

## Teaching example parameters, not a real drug

Dose = 100 mg, Volume (V) = 50 L, Clearance (CL) = 5 L per hour. These are teaching numbers only. They are not for any real drug and not for any real patient. Never use them to dose a person.

```python
import numpy as np
import matplotlib.pyplot as plt

# Teaching example parameters only, not a real drug
dose_mg = 100.0
V_L = 50.0
CL_L_per_h = 5.0
k_per_h = CL_L_per_h / V_L
C0_mg_per_L = dose_mg / V_L
half_life_h = np.log(2) / k_per_h
AUC_mg_h_per_L = dose_mg / CL_L_per_h
print('Elimination rate constant k:', round(k_per_h, 4), 'per hour')
print('Starting concentration C0:', round(C0_mg_per_L, 2), 'mg per L')
print('Half life:', round(half_life_h, 2), 'hours')
print('AUC:', round(AUC_mg_h_per_L, 2), 'mg hour per L')
```

```python
t = np.linspace(0, 48, 200)
C = C0_mg_per_L * np.exp(-k_per_h * t)
plt.figure()
plt.plot(t, C)
plt.xlabel('Time (hours)')
plt.ylabel('Concentration (mg per L)')
plt.title('One compartment IV bolus, teaching example')
plt.grid(True)
plt.show()
```
