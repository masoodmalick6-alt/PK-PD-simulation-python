# Stage 5: Parameter sensitivity and honest limitations

Now we change one parameter at a time and watch what happens. This shows which parameter matters most for the curve, and it also shows why a simple model must not be used to make real dosing decisions.

## Teaching example parameters, not a real drug

Base case: Dose 100 mg, V 50 L, CL 5 L per hour, ka 1.0 per hour, EC50 2 mg per L. Each test changes only one parameter. Teaching numbers only.

```python
import numpy as np
import matplotlib.pyplot as plt

def oral_curve(t, dose_mg=100.0, V_L=50.0, CL_L_per_h=5.0, ka_per_h=1.0, F=1.0):
    k_per_h = CL_L_per_h / V_L
    return (F * dose_mg * ka_per_h / (V_L * (ka_per_h - k_per_h))) * (np.exp(-k_per_h * t) - np.exp(-ka_per_h * t))

t = np.linspace(0, 48, 400)
plt.figure()
plt.plot(t, oral_curve(t), label='Base case')
plt.plot(t, oral_curve(t, CL_L_per_h=10.0), label='Clearance doubled')
plt.plot(t, oral_curve(t, V_L=100.0), label='Volume doubled')
plt.plot(t, oral_curve(t, ka_per_h=0.3), label='ka slower (0.3 per hour)')
plt.xlabel('Time (hours)')
plt.ylabel('Concentration (mg per L)')
plt.title('Sensitivity: one parameter changed at a time')
plt.legend()
plt.grid(True)
plt.show()

# EC50 sensitivity, using the Emax model
Emax = 100.0
C_range = np.linspace(0, 20, 200)
plt.figure()
for ec50 in [1.0, 2.0, 5.0]:
    plt.plot(C_range, Emax * C_range / (ec50 + C_range), label='EC50 = ' + str(ec50))
plt.xlabel('Concentration (mg per L)')
plt.ylabel('Effect (teaching units)')
plt.title('EC50 sensitivity, Emax model')
plt.legend()
plt.grid(True)
plt.show()
```

## Honest limitations, read this before you report anything

This project uses a one compartment model with first order absorption and elimination. Real bodies are more complex: distribution to tissues, protein binding, metabolism, kidney and liver function, age, weight, other medicines, and disease all change PK and PD. A simulated curve is not proof of how a real drug behaves, and it must never be used to recommend a dose for a real patient. If you want stronger evidence, you need real data, a validated model, and supervision from a qualified pharmacometrician or clinician.
