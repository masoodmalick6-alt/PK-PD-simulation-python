# Stage 3: Multiple dosing and steady state

Most medicines are given again and again. Each new dose adds to what is left from the last dose, so the concentration builds up until it reaches a steady pattern. That pattern is called steady state.

## Teaching example parameters, not a real drug

Dose = 100 mg every 12 hours, 10 doses, V = 50 L, CL = 5 L per hour, ka = 1.0 per hour, F = 1.0. Teaching numbers only. They are not a dosing recommendation for any real drug or patient.

```python
import numpy as np
import matplotlib.pyplot as plt

# Teaching example parameters only
dose_mg = 100.0
V_L = 50.0
CL_L_per_h = 5.0
ka_per_h = 1.0
F = 1.0
k_per_h = CL_L_per_h / V_L
tau_h = 12.0
n_doses = 10

def single_oral(t):
    t = np.where(t < 0, 0, t)
    return (F * dose_mg * ka_per_h / (V_L * (ka_per_h - k_per_h))) * (np.exp(-k_per_h * t) - np.exp(-ka_per_h * t))

t = np.linspace(0, tau_h * n_doses, 1200)
C_total = np.zeros_like(t)
for i in range(n_doses):
    C_total = C_total + single_oral(t - i * tau_h)

last_interval = t >= (n_doses - 1) * tau_h
C_last = C_total[last_interval]
print('Peak in the last interval:', round(float(np.max(C_last)), 2), 'mg per L')
print('Trough at the end of the last interval:', round(float(C_last[-1]), 2), 'mg per L')
print('Peak after the first dose only, for comparison:')
t1 = np.linspace(0, tau_h, 400)
print(round(float(np.max(single_oral(t1))), 2), 'mg per L')
```

```python
plt.figure()
plt.plot(t, C_total)
plt.xlabel('Time (hours)')
plt.ylabel('Concentration (mg per L)')
plt.title('Multiple oral doses, teaching example')
plt.grid(True)
plt.show()
```
