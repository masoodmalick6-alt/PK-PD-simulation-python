# Stage 4: PD link, the Emax model

PK tells us the concentration over time. PD tells us the effect at a given concentration. The Emax model is a simple way to link the two: effect rises with concentration, then levels off at a maximum.

## Teaching example parameters, not a real drug or a real effect

Emax = 100 effect units, EC50 = 2 mg per L. These are teaching numbers only. They do not describe any real drug, any real disease, or any real patient response.

```python
import numpy as np
import matplotlib.pyplot as plt

# Teaching example parameters only
Emax = 100.0
EC50_mg_per_L = 2.0

def emax_effect(C):
    return Emax * C / (EC50_mg_per_L + C)

C_range = np.linspace(0, 20, 200)
E_range = emax_effect(C_range)
print('Effect at EC50 (should be half of Emax):', round(float(emax_effect(EC50_mg_per_L)), 1))
print('Effect at 10 times EC50:', round(float(emax_effect(10 * EC50_mg_per_L)), 1))

plt.figure()
plt.plot(C_range, E_range)
plt.xlabel('Concentration (mg per L)')
plt.ylabel('Effect (teaching units)')
plt.title('Emax concentration effect curve, teaching example')
plt.grid(True)
plt.show()
```

```python
# Effect over time, using the oral PK curve from Stage 2
import numpy as np
import matplotlib.pyplot as plt

dose_mg = 100.0
V_L = 50.0
CL_L_per_h = 5.0
ka_per_h = 1.0
F = 1.0
k_per_h = CL_L_per_h / V_L
t = np.linspace(0, 48, 400)
C = (F * dose_mg * ka_per_h / (V_L * (ka_per_h - k_per_h))) * (np.exp(-k_per_h * t) - np.exp(-ka_per_h * t))
E = emax_effect(C)
plt.figure()
plt.plot(t, C, label='Concentration')
plt.plot(t, E, label='Effect')
plt.xlabel('Time (hours)')
plt.legend()
plt.title('Concentration and effect over time, teaching example')
plt.grid(True)
plt.show()
print('Peak effect:', round(float(np.max(E)), 1))
```
